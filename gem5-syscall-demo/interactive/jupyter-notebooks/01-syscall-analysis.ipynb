## Load and Parse Syscall Trace

First, let's parse a gem5 trace file to extract syscall information.

```python
def parse_trace_file(filename):
    """
    Parse gem5 trace file and extract syscall information.
    
    Returns: DataFrame with columns [tick, syscall, args, result]
    """
    syscalls = []
    
    with open(filename, 'r') as f:
        for line in f:
            # Match syscall invocation
            match = re.search(r'(\d+):\s+.*?syscall\s+(\w+)\s*\(([^)]*)', line)
            if match:
                tick = int(match.group(1))
                name = match.group(2)
                args = match.group(3)
                syscalls.append({
                    'tick': tick,
                    'syscall': name,
                    'args': args,
                    'result': None
                })
            
            # Match syscall return
            match = re.search(r'(\d+):\s+.*?returned\s+(-?\d+)', line)
            if match and syscalls:
                result = int(match.group(2))
                syscalls[-1]['result'] = result
    
    return pd.DataFrame(syscalls)

# Example: Load a trace file
# df = parse_trace_file('trace.txt')
# print(f"Loaded {len(df)} syscalls")
# df.head()
```

## Generate Report

Create a summary report of the syscall trace.

```python
def generate_report(df):
    """
    Generate a comprehensive syscall analysis report.
    """
    print("="*60)
    print("SYSCALL ANALYSIS REPORT")
    print("="*60)
    print()
    
    print(f"Total syscalls: {len(df)}")
    print(f"Unique syscalls: {df['syscall'].nunique()}")
    print(f"Simulation duration: {df['tick'].max() - df['tick'].min():,} ticks")
    print()
    
    # Top syscalls
    print("Top 5 syscalls:")
    for syscall, count in df['syscall'].value_counts().head(5).items():
        pct = (count / len(df)) * 100
        print(f"  {syscall:20} {count:5} ({pct:5.1f}%)")
    print()
    
    # Error analysis
    df_with_results = df[df['result'].notna()]
    errors = df_with_results[df_with_results['result'] < 0]
    print(f"Failed syscalls: {len(errors)} ({len(errors)/len(df_with_results)*100:.1f}%)")
    print()
    
    print("="*60)

# Example: Generate report
# generate_report(df)
```
